```python
#Lab Test
b = {101: "Python", 102: "Data Structures", 103: "AI", 104: "Machine Learning"}
a = {101, 102, 103, 104}
r = []
c = {}

def borrow(s, i):
    if i not in b:
        print("Book does not exist")
    elif i not in a:
        print("Book unavailable")
    else:
        a.remove(i)
        r.append((s, i, "Borrowed"))
        if s in c:
            c[s] = c[s] + 1
        else:
            c[s] = 1
        print("Book borrowed successfully")

def ret(s, i):
    if i not in b:
        print("Book does not exist")
    elif i in a:
        print("Book not borrowed")
    else:
        a.add(i)
        r.append((s, i, "Returned"))
        print("Book returned successfully")

def show():
    print("Available books:")
    for i in a:
        print(i, b[i])

    print("Borrowed books:")
    for i in b:
        if i not in a:
            print(i, b[i])

while True:
    print("\n1.Borrow")
    print("2.Return")
    print("3.Show books")
    print("4.Show records")
    print("5.Show counts")
    print("6.Exit")

    x = input("Enter choice: ")

    if x == "1":
        s = input("Student ID: ")
        i = int(input("Book ID: "))
        borrow(s, i)

    elif x == "2":
        s = input("Student ID: ")
        i = int(input("Book ID: "))
        ret(s, i)

    elif x == "3":
        show()

    elif x == "4":
        print(r)

    elif x == "5":
        print(c)

    elif x == "6":
        print("Returning to terminal..........")
        break

    else:
        print("Invalid choice")
```
